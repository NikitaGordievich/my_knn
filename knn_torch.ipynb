```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
data=load_iris()
x_train,x_v,y_train,y_v=train_test_split(data['data'],data['target'],test_size=0.2,random_state=42)

std_scal=StandardScaler()

x_train=std_scal.fit_transform(x_train)
x_v=std_scal.transform(x_v)

model=KNeighborsClassifier(n_neighbors=1)
model.fit(x_train,y_train)
y_pred=model.predict(x_v)
accur=accuracy_score(y_v,y_pred)
print('accuracy score:',accur)

```

```python
from matplotlib import pyplot as plt
fig=plt.figure()
ax=fig.add_subplot(projection='3d')
ax.scatter(*x_train[:,:3].T,cmap='viridis',c=y_train)
ax.scatter(*x_v[:,:3].T,cmap='cool',c=y_v,linewidths=1,edgecolors='black')
plt.show()
    
```
